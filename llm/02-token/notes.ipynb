# Counter函数验证

```python
from collections import Counter

# 1. 丢给它一个列表，它会自动帮你数数
my_counter = Counter(['苹果', '香蕉', '苹果', '橙子', '香蕉', '苹果'])
print(my_counter)  
# 输出: Counter({'苹果': 3, '香蕉': 2, '橙子': 1})

# 2. 访问不存在的键不会报错
print(my_counter['西瓜'])  
# 输出: 0 (普通字典这里就报错了)

# 3. 还可以直接查前几名（非常实用！）
print(my_counter.most_common(1))  
# 输出: [('苹果', 3)] (最常出现的前1名)
```

# split()函数验证

- 自动忽略开头和结尾的空格
- 如果不给 split() 传任何参数，它会默认以“空白字符”作为切刀来切分字符串。这里的空白字符包括：空格、换行符（\n）、制表符（\t）等
- 不管中间连续有多少个空格、换行符，它都会合并成一个来处理，不会切出空字符串。

```python
# 故意写得很乱的文本：有连续空格、有换行、有制表符
text2 = "  大模型   人工智能\n手写  \t算法  "
print(text2.split())
```

# 加载tokenizer-eval/universal-bpe数据集

```python
import os
import random

# 1. 依然配置好你的腾讯云环境
os.environ["HF_HOME"] = "/opt/datasets"
os.environ["HF_ENDPOINT"] = "https://hf-mirror.com"

from datasets import load_dataset

print("开始流式读取 quickmt-train.zh-en 数据集...")

# 2. 核心：加上 streaming=True 开启流式模式
dataset = load_dataset(
    "quickmt/quickmt-train.zh-en", split="train", streaming=True
)

english_texts = []
chinese_texts = []

# 3. 只读取前 10,000 行（一共可以拿到 10000条英文 + 10000条中文）
for i, item in enumerate(dataset):
    english_texts.append(item["en"])
    chinese_texts.append(item["zh"])
    if i >= 9999:
        break

print(f"✅ 成功提取英文语料: {len(english_texts)} 条")
print(f"✅ 成功提取中文语料: {len(chinese_texts)} 条")

# 4. 把它们混合并打乱，制成最终的 BPE 训练语料
final_corpus = english_texts + chinese_texts
random.seed(42)
random.shuffle(final_corpus)

print("\n" + "=" * 50)
print(f"🎉 BPE 语料库混合成功！可用文本总数: {len(final_corpus)} 行")
print("=" * 50)

# 5. 打印前两条看看数据成色
print("数据样例展示:")
print("[1]", final_corpus[0])
print("[2]", final_corpus[1])
```
