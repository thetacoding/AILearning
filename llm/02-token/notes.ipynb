# Counter函数验证

```python
from collections import Counter

# 1. 丢给它一个列表，它会自动帮你数数
my_counter = Counter(['苹果', '香蕉', '苹果', '橙子', '香蕉', '苹果'])
print(my_counter)  
# 输出: Counter({'苹果': 3, '香蕉': 2, '橙子': 1})

# 2. 访问不存在的键不会报错
print(my_counter['西瓜'])  
# 输出: 0 (普通字典这里就报错了)

# 3. 还可以直接查前几名（非常实用！）
print(my_counter.most_common(1))  
# 输出: [('苹果', 3)] (最常出现的前1名)
```

# split()函数验证

- 自动忽略开头和结尾的空格
- 如果不给 split() 传任何参数，它会默认以“空白字符”作为切刀来切分字符串。这里的空白字符包括：空格、换行符（\n）、制表符（\t）等
- 不管中间连续有多少个空格、换行符，它都会合并成一个来处理，不会切出空字符串。

```python
# 故意写得很乱的文本：有连续空格、有换行、有制表符
text2 = "  大模型   人工智能\n手写  \t算法  "
print(text2.split())
```
